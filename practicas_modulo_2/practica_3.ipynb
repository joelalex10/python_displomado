1. En un estudio reciente se descubrió que 90% de las familias tiene televisores
de pantalla grande. En una muestra de nueve familias, ¿cuál es la
probabilidad de que:\
a) las nueve tengan televisores de pantalla grande?\
b) menos de cinco tengan televisores de pantalla grande?\
c) más de cinco tengan televisores de pantalla grande?\
d) al menos siete familias tengan televisores de pantalla grande?

```python
from scipy.stats import binom

# parámetros
n = 9
p = 0.9

# a) las nueve tengan televisores
a = binom.pmf(9, n, p)

# b) menos de cinco tengan televisores
b = binom.cdf(4, n, p)

# c) más de cinco tengan televisores
c = 1 - binom.cdf(5, n, p)

# d) al menos siete tengan televisores
d = 1 - binom.cdf(6, n, p)

print("a) P(9) =", a)
print("b) P(<5) =", b)
print("c) P(>5) =", c)
print("d) P(>=7) =", d)
```

2. Un estudio reciente con respecto a salarios por hora de integrantes de
equipos de mantenimiento de las aerolíneas más importantes demostró que el
salario medio por hora era de $20.50, con una desviación estándar de $3.50.
Suponga que la distribución de los salarios por hora es una distribución de
probabilidad normal. Si elige un integrante de un equipo al azar, ¿cuál es la
probabilidad de que gane:\
a) entre 20.50 y 24.00 la hora?\
b) más de 24.00 la hora? \
c) menos de 19.00 la hora?

```python
from scipy.stats import norm

mu = 20.5
sigma = 3.5

# a) entre 20.50 y 24.00
a = norm.cdf(24, mu, sigma) - norm.cdf(20.5, mu, sigma)

# b) más de 24
b = 1 - norm.cdf(24, mu, sigma)

# c) menos de 19
c = norm.cdf(19, mu, sigma)

print("a) P(20.5 < X < 24) =", a)
print("b) P(X > 24) =", b)
print("c) P(X < 19) =", c)
```
