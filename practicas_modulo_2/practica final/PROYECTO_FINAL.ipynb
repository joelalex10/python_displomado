## 1. Carga el dataset de tu número de grupo de la plataforma del curso.

```python
import pandas as pd
import matplotlib.pyplot as plt

import os
```

```python

df = pd.read_csv("practica final/RetailNuevo2.csv")

df.head()
```

## 2. Expresa las ventas segmentadas por estado, por modo de entrega (Ship Mode) y por sub-categoría.

```python
ventasEstado = df.groupby('State')['Sales'].sum()
ventasEstado
```

```python
ventasShipMode = df.groupby('Ship Mode')['Sales'].sum()
ventasShipMode
```

```python
ventasSubCategoria = df.groupby('Sub-Category')['Sales'].sum()
ventasSubCategoria

conteoSubCategoria = df.groupby('Sub-Category')['Sales'].count()
conteoSubCategoria
```

## 3. Por cada categoría haz un diagrama de torta que exprese el porcentaje de ventas de cada sub categoría.


```python
# agrupar ventas por categoria y subcategoria
ventas = df.groupby(['Category','Sub-Category'])['Sales'].sum().reset_index()

# obtener lista de categorias
categorias = ventas['Category'].unique()

# generar grafico por categoria
for categoria in categorias:

    data_categoria = ventas[ventas['Category'] == categoria]

    plt.figure()

    plt.pie(
        data_categoria['Sales'],
        labels=data_categoria['Sub-Category'],
        autopct='%1.1f%%'
    )

    plt.title(f"Porcentaje de ventas por Subcategoría\nCategoría: {categoria}")

    plt.show()
```
