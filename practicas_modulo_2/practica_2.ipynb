1. El mes anterior la Asociación Nacional de Administradores de Salas
Cinematográficas realizó una encuesta entre 500 adultos seleccionados al
azar. La encuesta preguntaba a las personas su edad y el número de
películas que habían visto en el cine en el mes. La asociación está
interesada en entender las probabilidades de que un adulto vaya a ver una
película al cine especialmente en el caso de adultos mayores de 60 años.
Esta información es útil para tomar decisiones con respecto a los
descuentos en boletos y concesiones para los mayores
El resultado fue el siguiente:


![Tabla de datos](imagenes/image.png)


Determine la probabilidad de seleccionar un adulto que vio:

I. 6 o más películas por mes

II. 2 o menos películas por mes

III. 6 o más películas por mes o tiene 60 años o más

IV. 6 o más películas por mes dado que la persona tiene 60 años o más

V. 6 o más películas por mes y tiene 60 años o más

```python
import pandas as pd

data = {
    "peliculas": [0,1,2,3,4,5,6],
    "menor60": [50,60,70,40,30,20,10],
    "mayor60": [20,30,40,25,20,10,5]
}

df = pd.DataFrame(data)

df["total"] = df["menor60"] + df["mayor60"]

df
```

```python
total_personas = df["total"].sum()
print(total_personas)
```

```python
prob_6omas = df[df["peliculas"] >= 6]["total"].sum() / total_personas

print("P(6 o más películas) =", prob_6omas)
```

```python
prob_2omenos = df[df["peliculas"] <= 2]["total"].sum() / total_personas

print("P(2 o menos películas) =", prob_2omenos)
```

```python
personas_6omas = df[df["peliculas"] >= 6]["total"].sum()

personas_mayor60 = df["mayor60"].sum()

personas_6omas_y_mayor60 = df[df["peliculas"] >= 6]["mayor60"].sum()

prob_union = (personas_6omas + personas_mayor60 - personas_6omas_y_mayor60) / total_personas

print("P(6+ peliculas OR edad>=60) =", prob_union)
```

```python
prob_condicional = personas_6omas_y_mayor60 / personas_mayor60

print("P(6+ peliculas | edad>=60) =", prob_condicional)
```
