### Análisis Inicial de Datos (Machine Learning)

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

## PUNTO 1 · Carga del dataset

```python
df = pd.read_csv("Datos_pr.csv", encoding="latin1")
print(f"✔ Archivo cargado exitosamente.")
print(f"Dimensiones: {df.shape[0]} filas × {df.shape[1]} columnas")
```

## PUNTO 2 · Preguntas de la encuesta

```python
preguntas_originales = [c for c in df.columns if c not in ("Response Id", "Personality")]
for i, q in enumerate(preguntas_originales, 1):
    print(f"Q{i:02d}: {q}")
```

## PUNTO 3 · Traducción de columnas

```python
traduccion = {
    "Response Id": "ID_Respuesta",
    "Personality": "Personalidad"
}

df_trad = df.rename(columns=traduccion)
df_trad.head()
```

## PUNTO 4 · Guardar datos traducidos

```python
df_trad.to_csv("Datos_pr.csv", index=False, encoding="utf-8-sig")
print("Archivo guardado correctamente")
```

## PUNTO 5 · Número de registros

```python
n_registros = len(df_trad)
n_columnas = len(df_trad.columns)
n_preguntas = n_columnas - 2

print(f"Registros: {n_registros}")
print(f"Columnas: {n_columnas}")
print(f"Preguntas: {n_preguntas}")
```

## PUNTO 6 · Distribución de personalidad

```python
conteo = df_trad["Personalidad"].value_counts()

plt.figure()
plt.bar(conteo.index, conteo.values)
plt.xticks(rotation=45)
plt.title("Distribución de Personalidades")
plt.tight_layout()
plt.savefig("grafica_personalidades.png")
plt.show()
```

## PUNTO 7 · Descripción de personalidades

```python
descripcion_mbti = {
    "INTJ": "El Arquitecto",
    "ENTJ": "El Comandante"
}

for tipo in conteo.index:
    print(tipo, "-", descripcion_mbti.get(tipo, "Sin descripción"))
```
