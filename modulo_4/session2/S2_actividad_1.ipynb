### Parte B: Transición a Árboles - Fronteras No Lineales con Pipelines

Hasta ahora hemos observado  que un modelo estrictamente lineal falla al intentar aislar clases con topologías circulares. 
Antes de introducir los Árboles de Decisión, existe una técnica matemática para "doblar" el espacio: la **Transformación Polinomial**. 



Para aplicarla correctamente sin filtrar información (*data leakage*), utilizaremos la clase `Pipeline` de Scikit-Learn.

#### Guía Teórica y Sintaxis: El Flujo de Trabajo (Pipeline)
Un `Pipeline` es una herramienta que encadena múltiples pasos de procesamiento. Analicemos la siguiente estructura de código:

```python
model_pipe = Pipeline([
    ('poly', PolynomialFeatures(degree=2)),
    ('logit', LogisticRegression())
])
```
