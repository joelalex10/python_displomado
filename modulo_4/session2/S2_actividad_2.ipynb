En `scikit-learn`, instanciamos un árbol con la clase `DecisionTreeClassifier()`. Su hiperparámetro más crítico es **`max_depth`**, el cual dictamina cuántos niveles de divisiones puede realizar el modelo.

* **Profundidad = 2:** El árbol solo puede hacer 2 "preguntas" secuenciales. Generará fronteras de decisión cuadradas muy simples y masivas. Tiende a generalizar bien, pero podría sufrir de *Underfitting* si los datos son muy complejos.
